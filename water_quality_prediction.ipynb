```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv(r'C:\Users\legen\Downloads\afa2e701598d20110228.csv', sep=';')

print(df.columns)

cols_to_convert = df.columns.drop(['id', 'date'])

for col in cols_to_convert:
    df[col] = pd.to_numeric(df[col], errors='coerce')
    
df = df.dropna(subset=cols_to_convert)

print(df.shape)
print(df.dtypes)

plt.figure(figsize=(10, 6))
sns.heatmap(df[cols_to_convert].corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
